```python
# Initialize a dictionary to store outliers for each column
outliers_dict = {}

# Loop through each column in the DataFrame
for column in wine_df.columns:
    # Calculate IQR for the current column
    Q1 = wine_df[column].quantile(0.25)
    Q3 = wine_df[column].quantile(0.75)
    IQR = Q3 - Q1

    # Identify outliers for the current column
    outliers = wine_df[(wine_df[column] < (Q1 - 1.5 * IQR)) | (wine_df[column] > (Q3 + 1.5 * IQR))]
    outliers_dict[column] = outliers

    # Print outliers for the current column
    print(f"Outliers based on IQR for column [{column}]:\n", outliers)
    print (f"Q1 - 1.5 * IQR ->",(Q1-1.5*IQR))
    print (f"Q3 + 1.5 * IQR ->", (Q1+1.5*IQR) )
    print(f"Number of outliers in column [{column}]: {len(outliers)}\n")

```

```python
distortions = []
for k in range(1,6):
    kmeans = KMeans(n_clusters=k,random_state=42)
    kmeans.fit(wine_df_normalized)
    distortions.append(kmeans.inertia_)

plt.figure(figsize=(8,5))
plt.plot(range(1,6), distortions, marker ='o')

plt.xlabel('Number of clusters')
plt.ylabel('Distortion')
plt.title('Elbow method')
plt.show()
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
# How to find optimum number of cluster
# The K-means algorithm aims to choose centroids that minimise the inertia, or within-cluster sum-of-squares criterion:
```

```python
wcss=[]

for i in range(1,6):
    kmeans=KMeans(n_clusters=i)
    kmeans.fit(wine_df_normalized)
    wcss.append(kmeans.inertia_)

plt.plot(range(1,6),wcss)
plt.title('Elbow method')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.xticks(range(1,6))
plt.show()
```

```python
wcss
```

```python
# Low wcss: wcss are low and hence clusters are not scattered
```

```python
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i)
    kmeans.fit(wine_df_normalized)
    wcss.append(kmeans.inertia_)

plt.plot(range(1, 11), wcss, marker='o', linestyle='-', color='b')
plt.title('Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')

# Set x-axis ticks to display all values in the range (1, 11)
plt.xticks(range(1, 11))

# Annotate each point on the plot with its corresponding y-axis value
for i, txt in enumerate(wcss):
    plt.annotate(f'{txt:.2f}', (i+1, txt), textcoords="offset points", xytext=(0,10), ha='center')

# Highlight the points on the plot
plt.scatter(range(1, 11), wcss, c='r', marker='x')

plt.show()
```

# Comparison and Analysis

```python
The best number of clusters 'k values' was taken as 3 for original dataset, whereas for the PCA transformed data the optimum k value is 2.
The silhouette scores for both of them were around 0.6 which shows there was no significant changes after clustering.

Within cluster sum of squares values were in the range of 137-52 for the original dataset, but it is in the range of 17585049 - 1024764 for the PCA
transformed values.

-PCA Alters the Scale of Data

PCA transforms the original dataset into a new coordinate system (principal components). The new PCA components are a linear combination of original 
features, which can have larger numerical values. As a result, the sum of squared distances (WCSS) between points and cluster centroids increases 
significantly.

-PCA Can Change the Variance Range

In the original dataset, features may have been on a smaller scale (e.g., values between 0 and 100).
After PCA, the principal components capture variance, but their scales may be much larger.
The first few components might have very high variance, leading to large squared distances in WCSS.

-Impact of Standardization (if applied before PCA)

If you standardized (scaled) the data before PCA (e.g., using StandardScaler), the values would be centered around 0 with unit variance.
PCA creates new feature values that are not necessarily in the same scale as the original dataset.
This affects the WCSS calculation, leading to much larger numbers.

-PCA Does Not Affect Clustering Logic, Only Scale

Even though WCSS values differ, the relative trends remain the same (i.e., the elbow method still works).
The optimal number of clusters remains unchanged, even if the absolute values of WCSS increase.

WCSS increased drastically because PCA changed the feature scales. This is expected and does not necessarily mean clustering is worse. Instead of 
comparing absolute WCSS values, focus should be on the elbow point or silhouette scores for better evaluation.


Davies Bouldin index - this was 0.728 for the original dataset. It improved for the PCA transformed dataset which was 0.478. Lower value indicates better 
clusters.

Scatter plots show correlation in the original dataset. Clusters are scattered with some amount of overlapping. This could be visualized in the heatmap.
PCA tranformed data showed no correlation, clusters were clearly distinct and not overlapping. The scatter plot for original dataset showed that for
certain features like 'type','non-flavanoids','alcalinity','malic' the minimum value was Category 1 followed by Category 0 and the highest value was in
Category 2. But this trend was not followed for other features. When PCA transformation was performed, 0 Category had highest feature 'a' and negative
feature 'b'. Category 1 had negative Category 'a' and positive Category 'b'.

3D scatter plot - From the original dataset features like 'Type', ' Alcalinity' and 'Non-flavanoids' were taken and this formed clearly distinguishable 3 
clusters. The PCA transformed data showed majority of the data elements within one cluster.

The dendrogram also proved the same. In the original dataset there were 3 distinct clusters but the PCA transfomrd data showed 1 category. The other category
was not visible which could be because of its less percentage.

Based on the DB index, it can be concluded that the PCA transformed data gave better cluster results.
```

# Conclusion and insights

```python
Key findings and insights:-

Clustering effectiveness varied between original and PCA-transformed data.

PCA improved clustering performance by reducing noise and redundancy. This was seen by improved DB index.

After dimensionality reduction majority of the data elements stayed within one cluster.

Our analysis showed that PCA-based clustering resulted in well-separated clusters with a higher DB index. 
While PCA improved computational efficiency, some interpretability was lost due to transformed feature space.    

```
