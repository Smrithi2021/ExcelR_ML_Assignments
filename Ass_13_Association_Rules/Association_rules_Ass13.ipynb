```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
from mlxtend.preprocessing import TransactionEncoder

import matplotlib.pyplot as plt
```

```python
# Load the dataset (assuming excel format, were each row is a transaction)
retail_df=pd.read_excel("Online retail.xlsx",header=None)
```

```python
retail_df.head()
```

```python
# Convert each row into a list of items #Splitting by comma
```

```python
transactions=retail_df[0].str.split(",") 
```

```python
# transactions = retail_df.apply(lambda x: x.dropna().tolist(), axis=1).tolist()
```

```python
print(transactions)
```

```python
# Format for Apriori algorithm as above
```

```python
# Apriori algorith for association rule mining
```

```python
# Convert transactions into a format suitable for mlxtend
```

```python
te=TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)
df_transformed = pd.DataFrame(te_ary, columns =te.columns_)
```

```python
df_transformed
```

```python
df_transformed.value_counts()
```

```python
# Apply Apriori algorithm 
```

```python
frequent_itemsets= apriori(df_transformed, min_support=0.1, use_colnames=True)

```

```python
len(frequent_itemsets)
```

```python
frequent_itemsets
```

```python
# Generate association rules
```

```python
rules=association_rules(frequent_itemsets,metric='lift',min_threshold=1.0)

# Display results

print(rules)
```

```python
rules.sort_values('lift',ascending = False)
```

```python
# The min_support which was reduced to 0.1. In spite of that, the frequent items that appeared had single item sets. 
# Also, when the Association rule was applied there were no results shown. This means that data set does not have any
# set of items which were associated.
```

# 1.	What is lift and why is it important in Association rules?

```python
Why is Lift Important?
```

```python
Indicates True Association → Unlike support and confidence, lift accounts for how common individual items are.
    
Removes Random Co-occurrence → If items are bought frequently, confidence may be high even if they are unrelated. Lift corrects for this.

Helps Decision-Making → Businesses use lift to find valuable product combinations for recommendations and promotions.
```

```python
Lift Value	Meaning
> 1	Positive Association → Items are more likely to be bought together than by chance.
= 1	No Association → Items are bought together exactly as expected by chance.
< 1	Negative Association → Items are less likely to be bought together than by chance.
```
